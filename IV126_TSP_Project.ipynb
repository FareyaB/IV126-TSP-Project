```python

from pathlib import Path
import subprocess
import json
import time
import csv
import sys

# ------------------------------------
# PARAMETER TUNING CONFIGURATION
# ------------------------------------

removal_fractions = [0.10, 0.20]
acceptance_probabilities = [0.05, 0.10]

instance_file = "tsp_280.json"
test_timeout = 10

source = Path("main.py").read_text()

removal_original = "removal_count = max(1, n // 5)"
acceptance_original = "elif random.random() < 0.1:"

assert removal_original in source, "Removal parameter not found in main.py"
assert acceptance_original in source, "Acceptance parameter not found in main.py"

# Create a short-test instance
with open(instance_file) as f:
    instance = json.load(f)

instance["Timeout"] = test_timeout

Path("tuning_instance.json").write_text(
    json.dumps(instance)
)

matrix = instance["Matrix"]
n = len(matrix)

results = []

# ------------------------------------
# RUN PARAMETER EXPERIMENTS
# ------------------------------------

for fraction in removal_fractions:
    for probability in acceptance_probabilities:

        modified = source.replace(
            removal_original,
            f"removal_count = max(1, int(n * {fraction}))"
        ).replace(
            acceptance_original,
            f"elif random.random() < {probability}:"
        )

        Path("tuning_solver.py").write_text(modified)

        started = time.monotonic()
        valid = False
        distance = None
        status = "ERROR"

        try:
            result = subprocess.run(
                [
                    sys.executable,
                    "tuning_solver.py",
                    "tuning_instance.json",
                    "tuning_solution.json"
                ],
                capture_output=True,
                text=True,
                timeout=test_timeout + 10
            )

            if result.returncode == 0:
                route = json.loads(
                    Path("tuning_solution.json").read_text()
                )

                valid = (
                    isinstance(route, list)
                    and len(route) == n
                    and all(type(city) is int for city in route)
                    and sorted(route) == list(range(n))
                )

                if valid:
                    distance = sum(
                        matrix[route[i]][route[(i + 1) % n]]
                        for i in range(n)
                    )
                    status = "VALID"
                else:
                    status = "INVALID"
            else:
                status = "ERROR"
                print(result.stderr[-500:])

        except subprocess.TimeoutExpired:
            status = "TIMEOUT"

        except Exception as e:
            status = "ERROR"
            print("Error:", e)

        elapsed = round(time.monotonic() - started, 2)

        row = {
            "removal_fraction": fraction,
            "acceptance_probability": probability,
            "status": status,
            "distance": distance,
            "runtime": elapsed
        }

        results.append(row)

        print(
            f"Removal={fraction:.0%} | "
            f"Acceptance={probability:.0%} | "
            f"Status={status} | "
            f"Distance={distance} | "
            f"Runtime={elapsed}s",
            flush=True
        )

# ------------------------------------
# SAVE RESULTS
# ------------------------------------

with open("tuning_results.csv", "w", newline="") as f:
    writer = csv.DictWriter(
        f,
        fieldnames=[
            "removal_fraction",
            "acceptance_probability",
            "status",
            "distance",
            "runtime"
        ]
    )
    writer.writeheader()
    writer.writerows(results)

valid_results = [
    row for row in results
    if row["status"] == "VALID"
]

print("\n--- TUNING SUMMARY ---")

if valid_results:
    best = min(valid_results, key=lambda row: row["distance"])

    print("Best removal fraction:", best["removal_fraction"])
    print("Best acceptance probability:", best["acceptance_probability"])
    print("Best distance:", best["distance"])
else:
    print("No valid results.")

print("\nResults saved to tuning_results.csv")
```

```python

from pathlib import Path
import subprocess
import json
import csv
import sys
import time
import statistics

source = Path("main.py").read_text()

instance_file = "tsp_280.json"
test_timeout = 20
probabilities = [0.05, 0.10]
repetitions = 3

original = "elif random.random() < 0.1:"
assert original in source, "Acceptance setting not found"

with open(instance_file) as f:
    instance = json.load(f)

instance["Timeout"] = test_timeout
Path("repeat_tuning_instance.json").write_text(
    json.dumps(instance)
)

matrix = instance["Matrix"]
n = len(matrix)
results = []

for probability in probabilities:
    modified = source.replace(
        original,
        f"elif random.random() < {probability}:"
    )

    Path("repeat_tuning_solver.py").write_text(modified)

    for trial in range(1, repetitions + 1):
        output = Path("repeat_tuning_solution.json")
        output.unlink(missing_ok=True)

        start = time.monotonic()

        try:
            run = subprocess.run(
                [
                    sys.executable,
                    "repeat_tuning_solver.py",
                    "repeat_tuning_instance.json",
                    str(output)
                ],
                capture_output=True,
                text=True,
                timeout=test_timeout + 10
            )

            route = json.loads(output.read_text()) if (
                run.returncode == 0 and output.exists()
            ) else None

            valid = (
                isinstance(route, list)
                and len(route) == n
                and all(type(city) is int for city in route)
                and sorted(route) == list(range(n))
            )

            distance = (
                sum(
                    matrix[route[i]][route[(i + 1) % n]]
                    for i in range(n)
                )
                if valid else None
            )

            status = "VALID" if valid else "INVALID"

        except subprocess.TimeoutExpired:
            status = "TIMEOUT"
            distance = None

        elapsed = round(time.monotonic() - start, 2)

        results.append({
            "acceptance": probability,
            "trial": trial,
            "status": status,
            "distance": distance,
            "runtime": elapsed
        })

        print(
            f"Acceptance={probability:.0%}, "
            f"Trial={trial}, Status={status}, "
            f"Distance={distance}, Runtime={elapsed}s",
            flush=True
        )

with open("repeat_tuning_results.csv", "w", newline="") as f:
    writer = csv.DictWriter(
        f, fieldnames=results[0].keys()
    )
    writer.writeheader()
    writer.writerows(results)

print("\n--- AVERAGE DISTANCES ---")

for probability in probabilities:
    distances = [
        row["distance"] for row in results
        if row["acceptance"] == probability
        and row["status"] == "VALID"
    ]

    if distances:
        print(
            f"Acceptance {probability:.0%}: "
            f"Average={statistics.mean(distances):.2f}, "
            f"Best={min(distances):.2f}"
        )

print("\nSaved: repeat_tuning_results.csv")
```

```python
import json
import subprocess
import sys
from pathlib import Path

source = Path("main.py").read_text()
instance = json.loads(Path("tsp_666.json").read_text())

old_line = "elif random.random() < 0.1:"
assert old_line in source, "Could not find the current acceptance setting."

results = []

for probability in [0.05, 0.10]:
    modified = source.replace(
        old_line,
        f"elif random.random() < {probability}:"
    )

    solver_file = Path(f"tuning_666_{int(probability * 100)}.py")
    solver_file.write_text(modified)

    test_instance = dict(instance)
    test_instance["Timeout"] = 90

    instance_file = Path("tuning_666_instance.json")
    instance_file.write_text(json.dumps(test_instance))

    output_file = Path(f"tuning_666_solution_{int(probability * 100)}.json")
    if output_file.exists():
        output_file.unlink()

    print(f"Testing {probability:.0%} acceptance...", flush=True)

    run = subprocess.run(
        [sys.executable, str(solver_file), str(instance_file), str(output_file)],
        capture_output=True,
        text=True,
        timeout=105
    )

    if run.returncode != 0:
        print(run.stderr)
        raise RuntimeError("Solver failed")

    route = json.loads(output_file.read_text())
    n = len(instance["Matrix"])

    assert len(route) == n and set(route) == set(range(n)), "Invalid route"

    matrix = instance["Matrix"]
    distance = sum(
        matrix[route[i]][route[(i + 1) % n]]
        for i in range(n)
    )

    results.append((probability, distance))
    print(f"Distance: {distance:.2f}\n", flush=True)

print("FINAL COMPARISON")
for probability, distance in results:
    print(f"{probability:.0%} acceptance: {distance:.2f}")

print("Best in this test:", min(results, key=lambda x: x[1])[0])
```
